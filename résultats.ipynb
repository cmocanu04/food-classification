## 5. Analyse des résultats

LeNet-5 obtient une Accuracy de validation de **38.47 %** et un Macro F1 de **0.3471**. Cette expérience fournit une baseline légère pour la classification Food-11. LeNet-5 constitue un modèle peu coûteux en paramètres, mais ses performances de validation restent limitées, ce qui met en évidence l'intérêt du transfer learning et d'architectures plus profondes pour cette tâche.

L'entraînement de la seule tête de classification ne permet donc pas d'exploiter pleinement le potentiel de ResNet-18 pour cette tâche.
